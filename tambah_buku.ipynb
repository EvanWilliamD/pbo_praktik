```python
from google.colab import drive
drive.mount('/content/drive')

import sys
import os
import csv
import ipywidgets as widgets
from IPython.display import display

folder_path = "/content/drive/MyDrive/pbo_praktik"
file_path = os.path.join(folder_path, "buku.csv")

sys.path.append(folder_path)

from models import Buku


judul = widgets.Text(description="Judul:")
penulis = widgets.Text(description="Penulis:")
isbn = widgets.Text(description="ISBN:")

tombol = widgets.Button(
    description="Simpan Buku",
    button_style="primary"
)

output = widgets.Output()


def simpan_buku(b):
    with output:
        output.clear_output()

        if not judul.value or not penulis.value or not isbn.value:
            print("Semua kolom harus diisi!")
            return

        buku = Buku(judul.value, penulis.value, isbn.value)

        file_exists = os.path.exists(file_path)

        with open(file_path, "a", newline="", encoding="utf-8") as file:
            writer = csv.writer(file)

            if not file_exists or os.path.getsize(file_path) == 0:
                writer.writerow(["Judul", "Penulis", "ISBN", "Status"])

            writer.writerow(buku.to_list())

        print("Buku berhasil ditambahkan.")
        print("Lokasi file:", file_path)

        judul.value = ""
        penulis.value = ""
        isbn.value = ""


tombol.on_click(simpan_buku)

display(judul, penulis, isbn, tombol, output)
```
